```python
import pandas
import numpy
```

```python
dateDF = pandas.read_csv('https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day34-handling-date-and-time/orders.csv')
timeDF = pandas.read_csv('https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day34-handling-date-and-time/messages.csv')
```

```python
dateDF.head()
```

```python
timeDF.head()
```

```python
dateDF['date'] = pandas.to_datetime(dateDF['date'])
```

```python
dateDF['date_year'] = dateDF['date'].dt.year
dateDF['date_month'] = dateDF['date'].dt.month
dateDF['date_month_name'] = dateDF['date'].dt.month_name()
dateDF['date_day'] = dateDF['date'].dt.day
dateDF['date_day_name'] = dateDF['date'].dt.day_name()
```

```python
timeDF['date'] = pandas.to_datetime(timeDF['date'])
timeDF['date_year'] = timeDF['date'].dt.year
timeDF['date_month'] = timeDF['date'].dt.month
timeDF['date_month_name'] = timeDF['date'].dt.month_name()
```

```python
timeDF['date_day'] = timeDF['date'].dt.day
timeDF['date_day_name'] = timeDF['date'].dt.day_name()
```

```python
timeDF['date_hour'] = timeDF['date'].dt.hour
timeDF['date_minute'] = timeDF['date'].dt.minute
timeDF['date_second'] = timeDF['date'].dt.second
```
