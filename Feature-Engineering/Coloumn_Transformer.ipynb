# Column Transformer in Machine Learning using skLearn

```python
import pandas

df= pandas.read_csv("https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day28-column-transformer/covid_toy.csv")

df.head()
```

## Column Transformer Code

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder , StandardScaler , OrdinalEncoder

ct = ColumnTransformer(
    transformers=[
        ('Gender' , OneHotEncoder(sparse_output=False) , ['gender']),
        ('Output' , OrdinalEncoder(categories=[['No' , 'Yes']]) ,['has_covid']),
        ('Cough' ,OrdinalEncoder(categories=[['Mild' , 'Strong']]) ,['cough'])
    ],
    remainder='passthrough'
)

df_transformedData = ct.fit_transform(df)

df = pandas.DataFrame(df_transformedData)

df.head()
```
