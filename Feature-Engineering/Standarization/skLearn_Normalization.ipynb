```python
#skLearn-Normalization_Application
#sklearn wine datasets
import pandas as pd
import seaborn as sbs
from sklearn.datasets import load_wine
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

wine_datasets = load_wine(as_frame=True)
df = wine_datasets.data

# print(df[['alcohol' , 'proline']].head())

#Lets draw a graph using alcohol and proline
ax_before_graph = sbs.scatterplot(data=df , x='alcohol' , y='proline')
ax_before_graph.set_title('Before Standarization Transformation')
plt.show()

#Standarization
std_sklearn = StandardScaler()
df_standardized = pd.DataFrame(
    std_sklearn.fit_transform(df[['alcohol' , 'proline']]),
    columns=['alcohol' , 'proline'])

ax_after_graph = sbs.scatterplot(data=df_standardized , x='alcohol' , y='proline')
ax_after_graph.set_title('After Standarization Transformation')
plt.show()
```
