```python
import pandas

df = pandas.read_csv('https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day45-feature-construction-and-feature-splitting/train.csv')
df.sample(5)
```

## Applying Feature Construction

```python
#Create a new column in the df named family counts by summing the parch and sibsp and +1 and drop the both column after use
df['Family_count'] = df['Parch'] + df['SibSp'] + 1
# df.drop(columns=['Parch','SibSp'],inplace=True)
df.sample(5)
```

```python
#Now label the family counts based on large high and small frequency using label as 0, 1 ,2
def my_family_count(num):
    if num == 1:
        return 0
    elif num > 1 and num <= 4:
        return 1
    else:
        return 2
#Now Tranform using applying the apply function on the family count
df['Family_count'] = df['Family_count'].apply(my_family_count)
df.sample(5)
```
