# Multiple Linear Regression [dummy data]

###skLearn

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import numpy as np

x, y = load_diabetes(return_X_y=True)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

#Train a Linear Regression Model
lr = LinearRegression()
lr.fit(x_train, y_train)

y_pred = lr.predict(x_test)

#Evaluate the R2 Score
print(r2_score(y_test, y_pred))
```

### Own LinearRegression Class

```python
class mineLR():

  def __init__(self):
    self.coeff = None
    self.intercept = None

  def fit(self, x_train, y_train):
    x_train = np.insert(x_train, 0, 1, axis=1) #Add 1 to all
    #Start Applying the Formulae
    betas = np.linalg.inv(np.dot(x_train.T, x_train)).dot(x_train.T).dot(y_train)

    self.intercept = betas[0]
    self.coeff = betas[1:]

  def predict(self, x_test):
    y_pred = np.dot(x_test, self.coeff) + self.intercept
    return y_pred


mineLR = mineLR()
mineLR.fit(x_train, y_train)
y_pred = mineLR.predict(x_test)
print(r2_score(y_test, y_pred))
```
