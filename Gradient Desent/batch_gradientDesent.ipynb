## Batch Gradient Desent

##### OLS Formuale

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import numpy as np

diabetes = load_diabetes()
x = diabetes.data
y = diabetes.target

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

LR = LinearRegression()
LR.fit(x_train, y_train)
y_pred = LR.predict(x_test)

print(f'''R2Score : {r2_score(y_test, y_pred) * 100} %''')

print(f'''Slop : {LR.coef_}''')
print(f'''Intercept : {LR.intercept_}''')
```

## Custom Batch GD Class

```python
class BGDClass:
  def __init__(self , learning_rate , eipose):
    self.lr = learning_rate
    self.eipose = eipose
    self.coef = 0 # Beta 1,2,3 Its a 1D Array that stores all the values
    self.intersept = 1 #Beta zero it is

  def fit(self , X, Y):
    self.intersept = 0
    self.coef = np.ones(X.shape[1])

    for i in range(self.eipose):
      y_pred = np.dot(X, self.coef) + self.intersept
      intersept_derivative  = -2 * np.mean(Y - y_pred)
      coef_derivative = -2 * np.dot(X.T, (Y - y_pred)) / len(Y)

      self.intersept = self.intersept - (self.lr * intersept_derivative)
      self.coef = self.coef - (self.lr * coef_derivative)

  def predict(self , X):
    return np.dot(X, self.coef) + self.intersept

BGD = BGDClass(learning_rate=0.1, eipose=1500)
BGD.fit(x_train, y_train)
y_pred_bgd = BGD.predict(x_test)

print(f'''Slop : {BGD.coef}''')
print(f'''Intercept : {BGD.intersept}''')
print(f'''R2Score : {r2_score(y_test, y_pred_bgd) * 100} %''')
```
