## Mini Batch GD from scratch

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import numpy as np
import random

diabetes = load_diabetes()
x = diabetes.data
y = diabetes.target

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

LR = LinearRegression()
LR.fit(x_train, y_train)
y_pred = LR.predict(x_test)

print(f'''R2Score : {r2_score(y_test, y_pred) * 100} %''')

print(f'''Slop : {LR.coef_}''')
print(f'''Intercept : {LR.intercept_}''')
```

```python
class MBGDClass:
  def __init__(self , learning_rate , eipose , batch_size):
    self.batch_size = batch_size #Chosse a batch size of 20 or 30 or 50 or 100 according to dataset
    self.lr = learning_rate
    self.eipose = eipose
    self.coef = 0 # Beta 1,2,3 Its a 1D Array that stores all the values
    self.intersept = 1 #Beta zero it is

  def fit(self , X, Y):
    self.intersept = 0
    self.coef = np.ones(X.shape[1])

    for i in range(self.eipose):
        for j in range(int(X.shape[0] / self.batch_size)): #Runs equivalent to number of batches

            idx = random.sample(range(X.shape[0]), self.batch_size) #Selects a random batch of data from the dataset
                                                          #Exactly as the number of batch size we have given in the constructor
            
#Rather than using X ,Y we are using X[idx] and Y[idx] because we want to use only the batch data for the gradient descent
            y_pred = np.dot(X[idx], self.coef) + self.intersept

            intersept_derivative  = -2 * np.mean(Y[idx] - y_pred)
            coef_derivative = -2 * np.dot(X[idx].T, (Y[idx] - y_pred))  / len(Y[idx])

            self.intersept = self.intersept - (self.lr * intersept_derivative)
            self.coef = self.coef - (self.lr * coef_derivative)

  def predict(self , X):
    return np.dot(X, self.coef) + self.intersept

MBGD = MBGDClass(learning_rate=0.1, eipose=100 , batch_size=20)
MBGD.fit(x_train, y_train)
y_pred_bgd = MBGD.predict(x_test)

print(f'''Slop : {MBGD.coef}''')
print(f'''Intercept : {MBGD.intersept}''')
print(f'''R2Score : {r2_score(y_test, y_pred_bgd) * 100} %''')
```
